### Carregamento do BETA-dataset para o participante S001

```python
from scipy.io import loadmat
import numpy as np 

raw_data = loadmat('data/input/S1.mat')
raw_data.keys()
```

```python
import pandas as pd
import numpy as np

data = raw_data['data']
channels = pd.read_csv('data/input/64-channels.loc', sep=r'\s+', header=None, usecols=[3])[3].tolist() # Lê 64 canais do arquivo '64-channels.loc'
fs = 250

electrode_index, time_points, target_index, block_index = data.shape
print("Formato dos dados: ", data.shape, "\n")

time_vector = np.arange(time_points) / fs # Calcula quantas amostras

lista_dfs = []

for b in range(block_index):
    for t in range(target_index):
        trial_atual = data[:, :, t, b]

        columns_atual = {
            'block': b,
            'target': t,
            'time': time_vector
        }

        for i in range(electrode_index):
            columns_atual[channels[i]] = trial_atual[i, :]

        lista_dfs.append(pd.DataFrame(columns_atual))

df_long = pd.concat(lista_dfs, ignore_index=True)

print(df_long.head(10))
```

```python
numpy_array = df_long.to_numpy()
# Numpy Array
np.save('data/output/S001.npy', numpy_array)

# Array para o pré-processamento (numérico)
np.save('data/output/S001_pre_pros.npy', data)

numpy_array = np.load('data/output/S001.npy', allow_pickle=True)

print(numpy_array.shape)
print(numpy_array[:5])
```
