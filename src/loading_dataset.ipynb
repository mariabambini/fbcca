```python
# Carregamento do BETA-dataset para o participante S001

from scipy.io import loadmat
import numpy as np 

raw_data = loadmat('../data/S001.mat')
raw_data.keys()
```

```python
import pandas as pd

data = raw_data['data']

# Dimensões
print('Dimensões:', data.shape, '\n')

channels = ['POz', 'PO3', 'PO4', 'PO5', 'PO6', 'Oz', 'O1', 'O2']
electrodes = ['dry', 'wet']

fs = 250

# Formato dos dados
channels_index, time_points, electrode_index, block_index, target_index = data.shape

# Tempo de estimulação
pre_stimulus_time = 0.5
stimulus_time = 2
visual_latency = 0.14
post_stimulus_time = 0.2

# Parâmetros do trial
electrode = 0
block = 0
target = 0
qtd_rows = 10
first_sample = int((pre_stimulus_time + visual_latency) * fs)
final_sample = first_sample + qtd_rows

trial = data[:, :, electrode, block, target]

columns = {}
columns['time'] = np.arange(time_points) / fs - pre_stimulus_time

for i in range (channels_index):
    columns[channels[i]] = trial[i, :]

df = pd.DataFrame(columns)
df = df.iloc[first_sample:final_sample]

lista_dfs = []

for e in range(electrode_index):
    for b in range(block_index):
        for t in range(target_index):
            trial_atual = data[:, :, e, b, t]

            columns_atual = {}
            columns_atual['electrode'] = electrodes[e]
            columns_atual['block'] = b
            columns_atual['target'] = t
            columns_atual['time'] = np.arange(time_points) / fs - pre_stimulus_time

            for i in range(channels_index):
                columns_atual[channels[i]] = trial_atual[i, :]

            lista_dfs.append(pd.DataFrame(columns_atual))

df_long = pd.concat(lista_dfs, ignore_index=True)

print(df_long.head(10))
```

```python
numpy_array = df_long.to_numpy()
np.save('S001.npy', numpy_array)

numpy_array = np.load('S001.npy', allow_pickle=True)

print(numpy_array.shape)
print(numpy_array[:5])
```
